```python
# IMPORTAR LIBRERÍAS
import numpy as np
import matplotlib.pyplot as plt
```

# Distribución Weibull $We(\alpha,1)$

La función de distribución acumulada es:

$$
F(x)=1-e^{-x^\alpha}
$$

Aplicando el método de la transformada inversa:

$$
U=1-e^{-x^\alpha}
$$

Despejando:

$$
e^{-x^\alpha}=1-U
$$

$$
-x^\alpha=\ln(1-U)
$$

$$
x^\alpha=-\ln(1-U)
$$

Finalmente:

$$
X=(-\ln(U))^{1/\alpha}
$$

```python
# DISTRIBUCIÓN WEIBULL
alpha = 2
n = 10000

# Generar variables uniformes
U = np.random.uniform(0,1,n)

# Transformada inversa
X = (-np.log(U))**(1/alpha)

# Mostrar muestra
print("Primeros 10 valores:")
print(X[:10])

# Media
media = np.mean(X)
print("\nMedia:")
print(media)

# Varianza
varianza = np.var(X)
print("\nVarianza:")
print(varianza)

# Desviación estándar
desv = np.std(X)
print("\nDesviación estándar:")
print(desv)

# Histograma
plt.figure(figsize=(8,5))

plt.hist(X,
         bins=50,
         density=True)

plt.title("Histograma Weibull")
plt.xlabel("x")
plt.ylabel("Frecuencia")

plt.show()

# Muestra ordenada
X_ordenada = np.sort(X)

plt.figure(figsize=(8,5))

plt.plot(X_ordenada)

plt.title("Muestra ordenada Weibull")
plt.xlabel("Índice")
plt.ylabel("Valores")

plt.show()
```

# Distribución Gumbel $Gu(0,1)$

La función de distribución es:

$$
F(x)=e^{-e^{-x}}
$$

Aplicando transformada inversa:

$$
U=e^{-e^{-x}}
$$

$$
\ln(U)=-e^{-x}
$$

$$
-\ln(U)=e^{-x}
$$

$$
\ln(-\ln(U))=-x
$$

Por lo tanto:

$$
X=-\ln(-\ln(U))
$$

```python
# DISTRIBUCIÓN GUMBEL
n = 10000

U = np.random.uniform(0,1,n)

X = -np.log(-np.log(U))

print("Primeros 10 valores:")
print(X[:10])

media = np.mean(X)
varianza = np.var(X)
desv = np.std(X)

print("\nMedia:")
print(media)

print("\nVarianza:")
print(varianza)

print("\nDesviación estándar:")
print(desv)

# Histograma
plt.figure(figsize=(8,5))

plt.hist(X,
         bins=50,
         density=True)

plt.title("Histograma Gumbel")
plt.xlabel("x")
plt.ylabel("Frecuencia")

plt.show()

# Muestra ordenada
X_ordenada = np.sort(X)

plt.figure(figsize=(8,5))

plt.plot(X_ordenada)

plt.title("Muestra ordenada Gumbel")
plt.xlabel("Índice")
plt.ylabel("Valores")

plt.show()
```

# Distribución Cauchy $C(0,1)$

La función de distribución es:

$$
F(x)=\frac{1}{\pi}\arctan(x)+\frac12
$$

Aplicando transformada inversa:

$$
U=\frac{1}{\pi}\arctan(x)+\frac12
$$

$$
\pi\left(U-\frac12\right)=\arctan(x)
$$

Aplicando tangente:

$$
X=\tan\left[\pi\left(U-\frac12\right)\right]
$$

```python
# DISTRIBUCIÓN CAUCHY
n = 10000

U = np.random.uniform(0,1,n)

X = np.tan(np.pi*(U-0.5))

print("Primeros 10 valores:")
print(X[:10])

media = np.mean(X)
varianza = np.var(X)
desv = np.std(X)

print("\nMedia:")
print(media)

print("\nVarianza:")
print(varianza)

print("\nDesviación estándar:")
print(desv)

# Histograma
plt.figure(figsize=(8,5))

plt.hist(X,
         bins=100,
         density=True,
         range=(-10,10))

plt.title("Histograma Cauchy")
plt.xlabel("x")
plt.ylabel("Frecuencia")

plt.show()

# Muestra ordenada
X_ordenada = np.sort(X)

plt.figure(figsize=(8,5))

plt.plot(X_ordenada)

plt.title("Muestra ordenada Cauchy")
plt.xlabel("Índice")
plt.ylabel("Valores")

plt.show()
```

# Distribución Logística $L(0,1)$

La función de distribución es:

$$
F(x)=\frac{1}{1+e^{-x}}
$$

Aplicando transformada inversa:

$$
U=\frac{1}{1+e^{-x}}
$$

$$
\frac{1-U}{U}=e^{-x}
$$

Por lo tanto:

$$
X=\ln\left(\frac{U}{1-U}\right)
$$

```python
# DISTRIBUCIÓN LOGÍSTICA
n = 10000

U = np.random.uniform(0,1,n)

X = np.log(U/(1-U))

print("Primeros 10 valores:")
print(X[:10])

media = np.mean(X)
varianza = np.var(X)
desv = np.std(X)

print("\nMedia:")
print(media)

print("\nVarianza:")
print(varianza)

print("\nDesviación estándar:")
print(desv)

# Histograma
plt.figure(figsize=(8,5))

plt.hist(X,
         bins=50,
         density=True)

plt.title("Histograma Logística")
plt.xlabel("x")
plt.ylabel("Frecuencia")

plt.show()

# Muestra ordenada
X_ordenada = np.sort(X)

plt.figure(figsize=(8,5))

plt.plot(X_ordenada)

plt.title("Muestra ordenada Logística")
plt.xlabel("Índice")
plt.ylabel("Valores")

plt.show()
```

# Distribución Pareto $Par(\alpha)$

La función de distribución es:

$$
F(x)=1-x^{-\alpha}
$$

Aplicando transformada inversa:

$$
U=1-x^{-\alpha}
$$

$$
x^\alpha=\frac{1}{1-U}
$$

Por lo tanto:

$$
X=U^{-1/\alpha}
$$

```python
# DISTRIBUCIÓN PARETO
alpha = 3
n = 10000

U = np.random.uniform(0,1,n)

X = U**(-1/alpha)

print("Primeros 10 valores:")
print(X[:10])

media = np.mean(X)
varianza = np.var(X)
desv = np.std(X)

print("\nMedia:")
print(media)

print("\nVarianza:")
print(varianza)

print("\nDesviación estándar:")
print(desv)

# Histograma
plt.figure(figsize=(8,5))

plt.hist(X,
         bins=100,
         density=True,
         range=(1,5))

plt.title("Histograma Pareto")
plt.xlabel("x")
plt.ylabel("Frecuencia")

plt.show()

# Muestra ordenada
X_ordenada = np.sort(X)

plt.figure(figsize=(8,5))

plt.plot(X_ordenada)

plt.title("Muestra ordenada Pareto")
plt.xlabel("Índice")
plt.ylabel("Valores")

plt.show()
```
